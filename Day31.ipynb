```python
missing_percent = (df.isnull().sum() / len(df)) * 100

missing_data = pd.DataFrame({
    "Missing Values": df.isnull().sum(),
    "Missing Percentage": missing_percent
})

missing_data.sort_values(by="Missing Percentage", ascending=False)
```

```python
columns_to_fill = ["director", "country", "cast", "rating", "duration"]

for col in columns_to_fill:
    df[col] = df[col].fillna("Unknown")

df.isnull().sum()
```

```python
df["date_added"] = pd.to_datetime(df["date_added"], errors="coerce")

df["date_added"].isnull().sum()
```

```python
df.isnull().sum()
```

```python
content_counts = df["type"].value_counts()

print(content_counts)
```

```python
content_percentage = df["type"].value_counts(normalize=True) * 100

print(content_percentage.round(2))
```

```python
plt.figure(figsize=(8, 5))

sns.countplot(data=df, x="type")

plt.title("Distribution of Movies and TV Shows on Netflix")
plt.xlabel("Content Type")
plt.ylabel("Number of Titles")

plt.show()
```

```python
yearly_content = df["release_year"].value_counts().sort_index()

yearly_content.tail(15)
```

```python
plt.figure(figsize=(14, 6))

yearly_content.plot(kind="line", marker="o")

plt.title("Netflix Content Release Trends Over the Years")
plt.xlabel("Release Year")
plt.ylabel("Number of Titles")
plt.grid(True)
plt.show()
```

```python
peak_year = yearly_content.idxmax()
peak_count = yearly_content.max()

print("Peak Release Year:", peak_year)
print("Number of Titles:", peak_count)
```

```python
genre_data = df["listed_in"].str.split(", ").explode()

genre_data.head(10)
```

```python
genre_counts = genre_data.value_counts()

genre_counts.head(10)
```

```python
plt.figure(figsize=(12, 6))

genre_counts.head(10).plot(kind="bar")

plt.title("Top 10 Most Popular Genres on Netflix")
plt.xlabel("Genre")
plt.ylabel("Number of Titles")
plt.xticks(rotation=45, ha="right")

plt.tight_layout()
plt.show()
```

```python
most_popular_genre = genre_counts.idxmax()
genre_count = genre_counts.max()

print("Most Popular Genre:", most_popular_genre)
print("Number of Titles:", genre_count)
```

```python
country_data = df["country"].str.split(", ").explode()

country_counts = country_data.value_counts()

country_counts.head(10)
```

```python
plt.figure(figsize=(12, 6))

country_counts.head(10).plot(kind="bar")

plt.title("Top 10 Countries Producing Netflix Content")
plt.xlabel("Country")
plt.ylabel("Number of Titles")
plt.xticks(rotation=45, ha="right")

plt.tight_layout()
plt.show()
```

```python
top_country = country_counts.idxmax()
top_country_count = country_counts.max()

print("Top Content Producing Country:", top_country)
print("Number of Titles:", top_country_count)
```

```python
rating_counts = df["rating"].value_counts()

rating_counts
```

```python
plt.figure(figsize=(12, 6))

rating_counts.plot(kind="bar")

plt.title("Distribution of Netflix Content Ratings")
plt.xlabel("Rating")
plt.ylabel("Number of Titles")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
most_common_rating = rating_counts.idxmax()
rating_count = rating_counts.max()

print("Most Common Rating:", most_common_rating)
print("Number of Titles:", rating_count)
```

```python
movies = df[df["type"] == "Movie"].copy()

tv_shows = df[df["type"] == "TV Show"].copy()

print("Total Movies:", len(movies))
print("Total TV Shows:", len(tv_shows))
```

```python
movies["duration_minutes"] = (
    movies["duration"]
    .str.extract(r"(\d+)")
    .astype(float)
)

movies[["title", "duration", "duration_minutes"]].head()
```

```python
average_duration = movies["duration_minutes"].mean()
median_duration = movies["duration_minutes"].median()

print("Average Movie Duration:", round(average_duration, 2), "minutes")
print("Median Movie Duration:", median_duration, "minutes")
```

```python
plt.figure(figsize=(10, 6))

sns.histplot(movies["duration_minutes"].dropna(), bins=30, kde=True)

plt.title("Distribution of Netflix Movie Durations")
plt.xlabel("Duration (Minutes)")
plt.ylabel("Number of Movies")

plt.show()
```

```python
tv_shows["seasons"] = (
    tv_shows["duration"]
    .str.extract(r"(\d+)")
    .astype(float)
)

print("Average Seasons:", round(tv_shows["seasons"].mean(), 2))
print("Median Seasons:", tv_shows["seasons"].median())
```

```python
content_trends = df.groupby(["release_year", "type"]).size().unstack(fill_value=0)

content_trends.tail(15)
```

```python
plt.figure(figsize=(14, 6))

content_trends.plot(figsize=(14, 6))

plt.title("Movies vs TV Shows Release Trends")
plt.xlabel("Release Year")
plt.ylabel("Number of Titles")
plt.grid(True)
plt.tight_layout()

plt.show()
```

```python
df["year_added"] = df["date_added"].dt.year

added_year_counts = df["year_added"].value_counts().sort_index()

added_year_counts
```

```python
plt.figure(figsize=(12, 6))

added_year_counts.plot(kind="bar")

plt.title("Content Added to Netflix by Year")
plt.xlabel("Year Added")
plt.ylabel("Number of Titles")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```
